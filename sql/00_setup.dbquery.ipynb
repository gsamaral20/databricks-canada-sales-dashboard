```sql
-- Cria o catálogo do curso (só se ainda não existir)
CREATE CATALOG IF NOT EXISTS ai_bi_data_analyst;

-- Cria o schema onde ficarão as tabelas dos dashboards dentro do catálogo
CREATE SCHEMA IF NOT EXISTS ai_bi_data_analyst.ai_bi_dashboards;

-- Cria o volume onde ficarão os arquivos brutos (CSVs) do curso
CREATE VOLUME IF NOT EXISTS ai_bi_data_analyst.ai_bi_dashboards.source_files;
```
