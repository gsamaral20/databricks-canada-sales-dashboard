```sql
-- Confere a quantidade de linhas de cada tabela (esperado: 9, 40, 4, 15536, 342)
SELECT 
    'ca_products' AS tabela, 
    COUNT(*) AS linhas 
FROM ai_bi_data_analyst.ai_bi_dashboards.ca_products
UNION ALL
SELECT 
    'ca_customers',
    COUNT(*) 
FROM ai_bi_data_analyst.ai_bi_dashboards.ca_customers
UNION ALL
SELECT 'ca_sales_targets', 
    COUNT(*) 
FROM ai_bi_data_analyst.ai_bi_dashboards.ca_sales_targets
UNION ALL
SELECT 
    'ca_orders',        
    COUNT(*) 
FROM ai_bi_data_analyst.ai_bi_dashboards.ca_orders
UNION ALL
SELECT 
    'ca_opportunities', 
    COUNT(*) 
FROM ai_bi_data_analyst.ai_bi_dashboards.ca_opportunities;
```
